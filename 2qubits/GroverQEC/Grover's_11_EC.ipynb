```python
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit
import math
from qiskit.visualization import plot_distribution, plot_histogram
from qiskit_ibm_runtime import QiskitRuntimeService
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
from qiskit_ibm_runtime import SamplerV2 as Sampler
from qiskit_aer import AerSimulator
from qiskit_aer.noise import (
    NoiseModel,
    QuantumError,
    ReadoutError,
    depolarizing_error,
    pauli_error,
    thermal_relaxation_error,
)
```

```python
pi = math.pi
q = QuantumRegister(18,'q')
c = ClassicalRegister(2,'c')
qc = QuantumCircuit(q,c)
```

```python
QiskitRuntimeService.save_account(channel="ibm_quantum", token="Your Token Here", overwrite=True, set_as_default=True)
service = QiskitRuntimeService(channel='ibm_quantum')
backend = service.least_busy(min_num_qubits=127)

target = backend.target
pm = generate_preset_pass_manager(target=target, optimization_level=3)
circuit_qc = pm.run(qc)
noise_model = NoiseModel.from_backend(backend)
```

```python
qc.cx(q[0], q[6])
qc.cx(q[1], q[7])
qc.cx(q[0], q[12])
qc.cx(q[1], q[13])
qc.h(q[0])
qc.h(q[1])
qc.h(q[6])
qc.h(q[7])
qc.h(q[12])
qc.h(q[13])
qc.cx(q[0], q[2])
qc.cx(q[1], q[3])
qc.cx(q[0], q[4])
qc.cx(q[1], q[5])
qc.cx(q[6], q[8])
qc.cx(q[7], q[9])
qc.cx(q[6], q[10])
qc.cx(q[7], q[11])
qc.cx(q[12], q[14])
qc.cx(q[13], q[15])
qc.cx(q[12], q[16])
qc.cx(q[13], q[17])

qc.barrier()
```

Decoding

```python
qc.cx(q[0], q[2])
qc.cx(q[1], q[3])
qc.cx(q[0], q[4])
qc.cx(q[1], q[5])
qc.ccx(q[4], q[2], q[0])
qc.ccx(q[5], q[3], q[1])
qc.cx(q[6], q[8])
qc.cx(q[7], q[9])
qc.cx(q[6], q[10])
qc.cx(q[7], q[11])
qc.ccx(q[10], q[8], q[6])
qc.ccx(q[11], q[9], q[7])
qc.cx(q[12], q[14])
qc.cx(q[13], q[15])
qc.cx(q[12], q[16])
qc.cx(q[13], q[17])
qc.ccx(q[16], q[14], q[12])
qc.ccx(q[17], q[15], q[13])
qc.h(q[0])
qc.h(q[1])
qc.h(q[6])
qc.h(q[7])
qc.h(q[12])
qc.h(q[13])
qc.cx(q[0], q[6])
qc.cx(q[1], q[7])
qc.cx(q[0], q[12])
qc.cx(q[1], q[13])
qc.ccx(q[12], q[6], q[0])
qc.ccx(q[13], q[7], q[1])

qc.barrier()
```

Measure the logical qubits

```python
qc.measure(q[0], 0)
qc.measure(q[1], 1)
```

Draw

```python
qc.draw('mpl', fold=False, idle_wires=False)
```

Run on IBM Quantum Machine

```python
sampler = Sampler(mode=backend)
sampler.options.default_shots = 1_024
job = sampler.run([circuit_qc])
results = job.result()
```

Plot Results

```python
dist = results[0].data.c.get_counts()
plot_distribution(dist)
```

Run Simulator

```python
simulator = AerSimulator(noise_model=noise_model)
result = simulator.run(qc).result()
```

```python
counts = result.get_counts(qc)
plot_histogram(counts)
```
